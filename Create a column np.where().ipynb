# 1 Create a column based on 1 condition np.where()

```python
import numpy as np
```

```python
# create an array based on price tiers
# np.where(condition, condition is true, condition is false)
np.where(df_laptops['Price_euros'] > 2000,
         'Expensive', 'Cheap')
```

```python
# set it to a new column
df_laptops['Price_tier'] = np.where(df_laptops['Price_euros'] > 2000,
         'Expensive', 'Cheap')
```

```python
# show first 5 rows
df_laptops.head(5)
```

```python
# count values in price_tier column
df_laptops.value_counts('Price_tier')
```
