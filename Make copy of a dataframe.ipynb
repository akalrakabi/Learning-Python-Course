```python
import pandas as pd
```

```python
# read csv file
df = pd.read_csv('players_20.csv')
#set index
df.set_index('short_name', inplace=True)
#select column
df = df[['long_name', 'age', 'dob', 
         'height_cm', 'weight_kg', 
         'nationality','club']]
```

```python
df.head(3)
```

# copy()

## 1.1 deep=True 

```python
# deep=True by defualt (Modifications to the data or indices of the copy will not go to the original)

df_copy = df.copy()
```

```python
# update value in original dataframe
df.loc['L. Messi','height_cm'] = 180
```

```python
# copy vs original dataframe
print(df)
print('~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~')
print(df_copy)
```

## 1.2 deep = False

```python
# deep=False (Any changes to the data of original will be reflected in the shallow copy)

df_shallow_copy = df.copy(deep=False)
```

```python
df.loc['Cristiano Ronaldo', 'height_cm'] = 200
```

```python
# shallow copy vs original dataframe
print(df_shallow_copy)
print('~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~')
print(df)
```

## 1.3 Copy with a simple assginment 

```python
# make a copy
df_new_copy = df
```

```python
# update value in orignal dataframe
df.loc['Neymar Jr','height_cm'] = 190
```

```python
# copy vs original dataframe (Note, the copy becomes a shallow copy)
print(df)
print('~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~')
print(df_new_copy)
```
