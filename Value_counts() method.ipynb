```python
import pandas as pd
```

```python
# reading csv file
df_exams = pd.read_csv('StudentPerformance.csv')
df_exams.head()
```

# Value Counts

```python
# counting gender element

# len function (gives num of rows)
len(df_exams['gender'])

#.count() method
df_exams['gender'].count()
```

```python
# counting gender elements by category
df_exams['gender'].value_counts()
```

```python
# return the relative frequency or
# percentages (divide all values by the sum of values)
df_exams['gender'].value_counts(normalize=True)
```

```python
# counting 'parental leave' elements by category
df_exams['parental leave '].value_counts()
```

```python
# return the relative frequency and round 2 decimals
df_exams['parental leave '].value_counts(normalize=True).round(2)
```
