```python
# describing basic statistics of the dataframe
df_exams.describe()
# count is how many rows
# mean, sum of the data divided by the total rows (7)
# std, standard deviation
# min, minimum value
# max, max value
# 25%, Q1 percentile
# 50%, Q2
# 75%, Q3
```

```python
# obtaining the data type
type(df_exams)
```
