```python
import pandas as pd
```

```python
df_laptops = pd.read_csv('laptop_price.csv')
```

```python
df_laptops.head(3)
```

# 1 Filter dataframe based on one condition

```python
# find which rows have "Apple" in the "Company" column
df_laptops ['Company'] == "Apple"
```

```python
# filter dataframe based on condition
df_laptops [df_laptops ['Company'] == "Apple"]
```

```python
# verify which companies are in the "Company" column 
#with value_counts()
df_laptops[df_laptops ['Company'] == "Apple"].value_counts('Company')
```

## 1.1 Exercise 1

```python
# find which rows don't have "HP" in the "Company" column
df_laptops ['Company'] != "Apple"
```

```python
# filter dataframe based on condition
df_laptops [df_laptops ['Company'] != "HP"].value_counts('Company')
```
