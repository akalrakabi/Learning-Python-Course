```python
import pandas as pd
```

```python
# reading csv
df_cars = pd.read_csv('Car_sales.csv')
df_cars
```

```python
df_cars['Vehicle_type'].value_counts()
```

```python
# selecting columns
df_cars = df_cars[['Manufacturer','Sales_in_thousands',
         'Vehicle_type','Price_in_thousands', 
         'Engine_size','Horsepower','Fuel_capacity']]
```

```python
# show dataframe
df_cars
```

```python
# calculate number of unique values
df_cars.nunique()
```

# 1 agg()

```python
# calculate the sum of columns
df_cars_sample = df_cars[['Sales_in_thousands',
                          'Price_in_thousands', 
         'Engine_size','Horsepower','Fuel_capacity']]
df_cars_sample.agg('sum')
```

```python
# calculate the mean of numeric columns
df_cars_sample.agg('mean')
```

```python
# calculate the mean and count values
df_cars_sample.agg(['mean','count'])
```

```python
# Different aggregations per column: calculate the sum and mean of 'Sales_in_thousands'
# and sum and max of 'Price_in_thousands'
df_cars_sample.agg({'Sales_in_thousands':['sum', 'mean'],
                    'Price_in_thousands':['sum','max']})
```

```python
# aggregate over (along) the columns
df_cars_sample[['Sales_in_thousands','Price_in_thousands']].agg('sum', axis=1)
```

```python
# equivalent (NaN behaves differently)
df_cars_sample['Sales_in_thousands'] + df_cars_sample['Price_in_thousands']
```

Note: agg() treats NaN as zero

```python
# aggregate different functions and 
# rename the index of the resulting dataframe
df_cars_sample.agg(x = ('Sales_in_thousands','sum'),
                   y = ('Price_in_thousands','sum'))
```

# 2 The Split-Apply-Combine Strategy

```python
# split data into separate groups

passenger_filter = df_cars['Vehicle_type']=='Passenger'
car_filter = df_cars['Vehicle_type']=='Car'

#df_cars['Vehicle_type'].value_counts()
```

```python
# apply an operation
passenger_average = df_cars[passenger_filter]['Sales_in_thousands'].mean()
car_average = df_cars[car_filter]['Sales_in_thousands'].mean()
```

```python
# combine results
pd.DataFrame({'Vehicle_type':['Car','Passenger'],
              'Sales_in_thousands':[passenger_average,car_average]})
```

# 4 Groupby() and agg()

```python
# find the minimum and maximum values on each column
df_cars.groupby('Vehicle_type').min()
df_cars.groupby('Vehicle_type').max()
df_cars.groupby('Vehicle_type').agg(['min','max'])
```

```python
# set a name for aggregated column
# find the min Engine_size and max Horsepower
df_cars.groupby('Vehicle_type').agg(min_engine_size=('Engine_size', 'min'),
                                    max_horsepower=('Horsepower', 'max'))
```

```python
# calculate the sum of Sales_in_thousands and mean of Price_in_thousands
df_cars.groupby('Manufacturer').agg(sum_sales = ('Sales_in_thousands','sum'),
                                   mean_price = ('Price_in_thousands','mean'))
```

```python
df_cars.groupby('Manufacturer').sum()['Sales_in_thousands']
```

```python
df_cars.groupby('Manufacturer').mean()['Price_in_thousands']
```

# 5 Lambda

apply your own function

```python
sum_one = lambda x:x+1
sum_one(1)
```

```python
# get actual sales and price values
df_cars.groupby('Manufacturer').sum().apply(lambda x:x*1000)[['Sales_in_thousands','Price_in_thousands']]
```

```python
# subtract values from the mean of each group
df_cars_sample2 = df_cars[['Manufacturer','Sales_in_thousands',
                          'Price_in_thousands', 
         'Engine_size','Horsepower','Fuel_capacity']]
df_cars_sample2.groupby('Manufacturer').apply(lambda x:x-x.mean())
```

```python
# filter certain values out based on sum() and return a subset
def filter_func(x):
    return x['Sales_in_thousands'].sum() > 52

df_filter = df_cars.groupby(['Manufacturer']).filter(filter_func)
```
