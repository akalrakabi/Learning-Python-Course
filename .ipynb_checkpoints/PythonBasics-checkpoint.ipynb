### 6 If Statement

```python
if <condition>:
    <code>
elif <condition>:
    <code>
...
else:
    <code>
```

```python
age = 15

if age>=18:
    print("You are an adult")
elif age>=13:
    print("You are a teenager")
else:
    print("You are a child")
```

### 7 For Loop

```python
for <variable> in <iterable or list>:
    <code>
```

```python
countries
```

```python
for country in countries:
    print(country)
```

```python
for i, country in enumerate(countries):
    print(i,country)
```

```python
my_data = {'name':'Awfa','age':'28','height':'1.67'}
```

```python
my_data
```

```python
for key, value in my_data.items():
    print(key,value)
```
