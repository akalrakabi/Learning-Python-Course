```python
import pandas as pd
```

```python
# read csv file
df = pd.read_csv('players_20.csv')
#set index
df.set_index('short_name', inplace=True)
#select column
df = df[['long_name', 'age', 'dob', 
         'height_cm', 'weight_kg', 
         'nationality','club']]
```

```python
df.head(3)
```

# 1 Drop Row

```python
# drop one row
# axis parameter, 0 is rows and 1 is columns
df.drop('L. Messi', axis=0)
#or
# index parameter
df.drop(index=['L. Messi'])
```

```python
# drop two or more rows and update data (inplace=True)
df.drop(['L. Messi', 'Cristiano Ronaldo'], axis=0, inplace=True)
```

```python
df
```

# 2 Drop Columns 

```python
# drop one column
# axis parameter
df.drop('long_name', axis=1)
# or columns parameter
df.drop(columns=['long_name'])
```

```python
df
```

```python
# drop column by position/index (drop last column, club)
# df.columns[[6]]
# or
# df.columns[[-1]]
df.drop(df.columns[[-1]], axis=1)
```

```python
# drop 2 or more columns and update data (inplace=True)
df.drop(['long_name','dob'], axis=1, inplace=True)
```
