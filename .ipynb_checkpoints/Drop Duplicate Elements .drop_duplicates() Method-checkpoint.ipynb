```python
import pandas as pd
```

```python
df_laptops = pd.read_csv('laptop_price.csv')
```

```python
df_laptops.head(3)
```

# 1 drop_duplicated()

```python
# drop duplicates in 2 or more columns
df_laptops.drop_duplicates(['Company'])[['Company', 'Price_euros']].value_counts()
```

```python
# sort dataframe ascending by company and price
# cheapest first and most expensive last
df_laptops = df_laptops.sort_values(['Company','Price_euros'])
```

```python
# cheapest keep='first'
df_laptops.drop_duplicates(['Company'])[['Company','Price_euros']]
```

```python
# most expensive keep='last'
# save changes to dataframe inplace=True
df_laptops.drop_duplicates(['Company'], keep='last', inplace=True, ignore_index=True)
```

```python
# Arguments

# inplace: drop duplicates in place or return a copy.
# ignore_index=True: the resulting axis will be labeled 0,1,...
df_laptops[['Company', 'Price_euros']]
```

# 2 Exercise 1

Task: Find the biggest and smallest screen size in laptops of each company using the sort_values() 
and duplicated() methods (keep "first", "last", and False

```python
# Sort dataframe ascending by company and inches
# smallest first and biggest last
df_laptops = df_laptops.sort_values(['Company','Inches'])
```

```python
# smallest: keep='first'
df_laptops.drop_duplicates(['Company'])[['Company','Inches']]
```

```python
#biggest: keep='last'
df_laptops.drop_duplicates(['Company'], keep='last')[['Company','Inches']]
```
