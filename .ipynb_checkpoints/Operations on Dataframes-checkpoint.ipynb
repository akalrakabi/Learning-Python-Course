```python
import pandas as pd
```

```python
# reading the csv file
df_exams = pd.read_csv('StudentPerformance.csv')
df_exams.head()
```

# 1 Math Operations

## 1.1 Operations in columns

```python
# Select a column and calculate total sum
df_exams['math score'].sum()
```

```python
# count, mean, std, max, and min
print("count: ", df_exams['math score'].count())
print("mean: ",df_exams['math score'].mean())
print("std: ",df_exams['math score'].std())
print("max: ",df_exams['math score'].max())
print("min: ",df_exams['math score'].min())
```

```python
# easier calculation with .describe()
df_exams.describe()
```

## 1.2 Operations in rows

```python
# calculating the sum in a row
df_exams['math score'] + df_exams['reading score'] + df_exams['writing score']
```

```python
# calculating the average score & assigning result to new column
df_exams['average'] = (df_exams['math score'] + df_exams['reading score'] + df_exams['writing score'])/3
```

```python
# showing the dataframe
df_exams.round(2)
```
