```python
import pandas as pd
```

```python
# reading the csv file
df_exams = pd.read_csv('StudentPerformance.csv')
df_exams.head()
```

# 1 Selecting 2 or more columns

```python
# selecting 2 columns using [[]]
df_exams[['gender', 'math score']]
```

```python
# check out the data type of the selection
type(df_exams[['gender', 'math score']])
```

```python
# select 2 or more columns using [[]]
df_exams[['gender','math score', 'reading score','writing score']]
# Note: 1 pair of [] is a series (array)
#       2 pair of [[]] is a DataFrame
```

```python
# we can't select 2 or more columns with .
df_exams.'gender', 'math score'
```
