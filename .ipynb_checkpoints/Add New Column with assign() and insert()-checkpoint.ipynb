```python
import pandas as pd
```

```python
# reading the csv file
df_exams = pd.read_csv('StudentPerformance.csv')
df_exams.head()
```

# 1 Adding a new column

## 1.1 assign()

##### When to use assign?
##### > Add multiple columns in a single line of code
##### > When you need to overwrite the values of existing columns (best practice)
##### 
##### It returns a new object (a copy) with all the original columns in addition to the new ones

```python
import numpy as np
```

```python
# create random numbers for our new score columns
score1 = np.random.randint(1, 100, size=7)
score2 = np.random.randint(1, 100, size=7)
```

```python
print("score1: ", score1)
print("score2: ", score2)
```

```python
# create a series using the random numbers
series1 = pd.Series(score1, index=np.arange(0,7))
series2 = pd.Series(score2, index=np.arange(0,7))
```

```python
print("Series1: ", series1)
print("Series2: ", series2)
```

```python
# using assign() to add multiple columns (Note this is only a copy)
df_exams.assign(score1 = series1, score2 = series2)
```

```python
# to change the original dataframe, since assign only creates a copy
df_exams = df_exams.assign(score1 = series1, score2 = series2)
df_exams
```

# 1.2 insert()

#### inserts a new column at a specific position or index.

```python
# using index() to add a column at a specific position. 
# Note insert method does not return a copy, it overwirtes the dataframe
# insert(index number, 'name', value)
df_exams.insert(1, 'test', series1)
```

```python
# show dataframe
df_exams
```
