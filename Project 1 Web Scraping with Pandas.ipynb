# 2 Read CSV from Multiple URLs w/ Pandas

Link: root + season + league

https://football-data.co.uk/mmz4281/2425/E0.csv

```python
# Link structure
"https://football-data.co.uk/mmz4281/" + "2122" + "/" + "E0" + ".csv"
```

```python
# create a root variable
root = "https://football-data.co.uk/mmz4281/"
```

## 2.1 Multiple Leagues

```python
# creating list of Leagues
leagues = ['E0', 'E1', 'E2', 'E3', 'EC']
# empty list
frames = []

# looping through Leagues, read mulitple csv & append it into a list
for league in leagues:
    df = pd.read_csv(root + '2122/' + league + '.csv', encoding='unicode_escape')
    # store all dataframes in frames list, b/c in each iteration df refreshes
    frames.append(df)
```

```python
# length of frames
len(frames)
```

```python
# show 1st, 2nd, 3rd element
#frames[0]
# frames[1]
frames[2]
```

## 2.2 Multiple Seasons 

```python
# creating list of Leagues
leagues = ['E0', 'E2','E3']
# empty list
frames = []

# loop through range of seasons and insert new column
for league in leagues:
    # seasons 2015-2020
    for season in range(15, 21):
        df = pd.read_csv( root + str(season) + str(season+1) + '/' + league + '.csv')
        
        # adding season column for each dataframe
        df.insert(1, 'season', season)
        
        # store all dataframes in frames list, b/c in each iteration df refreshes
        frames.append(df)
```

```python
# total frames (6 seasons x 3 leagues)
len(frames)
```

```python
# show 1st & last element (new season column was created)
print(frames[0])
print('~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~')
print(frames[17])
```

# 3 Organize All the Data in a Dictionary

```python
# create a dictionary with original name of league as key
dict_countries = {'Spanish La Liga':'SP1', 'Spanish Segundo Division':'SP2',
                 'German Bundesliga':'D1',
                 'English Premier League': 'E0', 'English League 1': 'E2', 'English League 2':'E3'}
```

```python
# get a dictionary element
dict_countries['Spanish La Liga']
```

```python
for league in dict_countries:
    print(league, '-' ,dict_countries[league])
```

```python
dict_historical_data = {}

# creating list of Leagues
#leagues = ['E0', 'E2','E3']
# empty list
#frames = [] placing this inside the for loop so that 
#            the list doesn't contain older data after every iteration

# looping through key elements
for league in dict_countries:
    frames = []
    # seasons 2015-2020
    for season in range(15, 21):
        df = pd.read_csv( root + str(season) + str(season+1) + '/' + dict_countries[league] + '.csv')
        
        # adding season column for each dataframe
        df.insert(1, 'season', season)
        
        # store all dataframes in frames list, b/c in each iteration df refreshes
        frames.append(df)
    df_concat = pd.concat(frames)
    # df_concat is the value which contains all the dataframes from season 15-20, league is the key
    dict_historical_data[league] = df_concat
```
