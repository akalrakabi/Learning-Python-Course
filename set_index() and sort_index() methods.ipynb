```python
import pandas as pd
```

```python
# read csv
df_exams = pd.read_csv("StudentPerformance.csv")
df_exams
```

# 1 Create Index

```python
import numpy as np
import random
```

```python
# creating non repetitive values for the index
new_index = np.arange(0,7)
```

```python
# shuffling the indexes
random.shuffle(new_index)
```

```python
# create a new column with new index
df_exams['new_index'] = new_index
```

```python
df_exams
```

# 2 Set Index

```python
# set new_index column as index
# Note inplace argument dictates whether dataframe is updated, True means update
df_exams.set_index('new_index', inplace=True)
df_exams
```

# 3 Sort Index

```python
# sort dataframe by new index
df_exams.sort_index()
```

```python
# sort descending by index
df_exams.sort_index(ascending=False, inplace=True)
```
